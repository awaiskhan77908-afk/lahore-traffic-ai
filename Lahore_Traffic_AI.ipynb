```python

import pandas as pd
import numpy as np

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error
import joblib

# Load the demonstration dataset
df = pd.read_csv("/content/lahore_traffic_demo.csv")

df["timestamp"] = pd.to_datetime(df["timestamp"])
df = df.sort_values(["road_id", "timestamp"]).copy()

# Create time features
df["hour"] = df["timestamp"].dt.hour
df["day_of_week"] = df["timestamp"].dt.dayofweek
df["minute_of_day"] = (
    df["timestamp"].dt.hour * 60
    + df["timestamp"].dt.minute
)

# Previous observation's speed for the same road
df["previous_speed"] = (
    df.groupby("road_id")["average_speed_kmh"].shift(1)
)

# Target: speed at the next 15-minute observation
df["target_speed"] = (
    df.groupby("road_id")["average_speed_kmh"].shift(-1)
)

df = df.dropna().copy()

features = [
    "hour",
    "day_of_week",
    "minute_of_day",
    "previous_speed"
]

X = df[features]
y = df["target_speed"]

# Chronological train/test split
split_index = int(len(df) * 0.8)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]
y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]

# Train the AI model
model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    min_samples_leaf=3
)

model.fit(X_train, y_train)

# Evaluate on held-out observations
predictions = model.predict(X_test)

mae = mean_absolute_error(y_test, predictions)

print("AI model trained successfully!")
print("Training observations:", len(X_train))
print("Testing observations:", len(X_test))
print("Test MAE:", round(mae, 2), "km/h")

# Save the model
joblib.dump(
    {"model": model, "features": features},
    "/content/traffic_model.joblib"
)

print("Model saved successfully!")
```

```python

%%writefile app.py

import streamlit as st
import pandas as pd
import joblib

st.set_page_config(
    page_title="Lahore Traffic AI",
    page_icon="🚦",
    layout="wide"
)

st.title("🚦 Lahore Traffic Congestion Prediction")
st.caption("AI-powered traffic forecasting — student FYP prototype")

st.warning(
    "Demo mode: predictions use synthetic data, "
    "not real-time Lahore traffic."
)

@st.cache_data
def load_data():
    data = pd.read_csv("/content/lahore_traffic_demo.csv")
    data["timestamp"] = pd.to_datetime(data["timestamp"])
    return data

@st.cache_resource
def load_model():
    return joblib.load("/content/traffic_model.joblib")

try:
    df = load_data()
    saved = load_model()
    model = saved["model"]
    features = saved["features"]

    road_name = st.selectbox(
        "Select a road",
        sorted(df["road_id"].unique())
    )

    road_data = df[df["road_id"] == road_name].sort_values("timestamp")
    latest = road_data.iloc[-1]

    next_time = latest["timestamp"] + pd.Timedelta(minutes=15)

    prediction_input = pd.DataFrame([{
        "hour": next_time.hour,
        "day_of_week": next_time.dayofweek,
        "minute_of_day": next_time.hour * 60 + next_time.minute,
        "previous_speed": latest["average_speed_kmh"]
    }])[features]

    predicted_speed = float(model.predict(prediction_input)[0])

    def congestion_level(speed):
        if speed < 15:
            return "Severe"
        elif speed < 25:
            return "High"
        elif speed < 35:
            return "Moderate"
        return "Low"

    level = congestion_level(predicted_speed)

    st.subheader("Traffic forecast")
    col1, col2, col3 = st.columns(3)

    col1.metric(
        "Last demo speed",
        f"{latest['average_speed_kmh']:.1f} km/h"
    )
    col2.metric(
        "Predicted speed",
        f"{predicted_speed:.1f} km/h"
    )
    col3.metric("Congestion level", level)

    st.info(
        f"Forecast target: {next_time:%d %b %Y, %H:%M}"
    )

    st.subheader("Historical demonstration speeds")

    chart_data = road_data.set_index("timestamp")[
        "average_speed_kmh"
    ]
    st.line_chart(chart_data)

    st.subheader("Dataset preview")
    st.dataframe(road_data.tail(10), use_container_width=True)

    st.download_button(
        "Download road data as CSV",
        data=road_data.to_csv(index=False),
        file_name="traffic_road_data.csv",
        mime="text/csv"
    )

except FileNotFoundError:
    st.error(
        "Required data or model file is missing. "
        "Create the dataset and train the model first."
    )
except Exception as error:
    st.error(f"Application error: {error}")
```

```python

import pandas as pd
import numpy as np

np.random.seed(42)

roads = [
    "Canal Road",
    "Ferozepur Road",
    "Mall Road",
    "Main Boulevard Gulberg",
    "Multan Road"
]

timestamps = pd.date_range(
    start="2026-01-01",
    periods=7 * 24 * 4,
    freq="15min"
)

rows = []

for road in roads:
    base_speed = np.random.randint(25, 45)

    for timestamp in timestamps:
        hour = timestamp.hour
        rush_hour = hour in [8, 9, 17, 18, 19]

        speed = base_speed + np.random.normal(0, 5)

        if rush_hour:
            speed -= np.random.randint(8, 18)

        speed = max(5, min(70, speed))

        rows.append({
            "timestamp": timestamp,
            "road": road,
            "speed": round(speed, 2)
        })

df = pd.DataFrame(rows)
df.to_csv("/content/lahore_traffic_demo.csv", index=False)

print("Dataset created!")
print("Total rows:", len(df))
print(df.head())
```

```python

import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error
import joblib
import os

# Dataset load karein
df = pd.read_csv("/content/lahore_traffic_demo.csv")
df["timestamp"] = pd.to_datetime(df["timestamp"])

# Time features banayein
df["hour"] = df["timestamp"].dt.hour
df["day_of_week"] = df["timestamp"].dt.dayofweek
df["road_id"] = df["road"].astype("category").cat.codes

# Har road ki previous speed aur next speed
df = df.sort_values(["road", "timestamp"]).copy()
df["previous_speed"] = df.groupby("road")["speed"].shift(1)
df["target_speed"] = df.groupby("road")["speed"].shift(-1)
df = df.dropna().copy()

features = [
    "hour", "day_of_week", "road_id", "previous_speed"
]
X = df[features]
y = df["target_speed"]

# Chronological split
split = int(len(df) * 0.8)
X_train, X_test = X.iloc[:split], X.iloc[split:]
y_train, y_test = y.iloc[:split], y.iloc[split:]

model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)
model.fit(X_train, y_train)

predictions = model.predict(X_test)
mae = mean_absolute_error(y_test, predictions)

# Model aur feature information save karein
joblib.dump({
    "model": model,
    "features": features,
    "road_codes": {
        road: int(code)
        for road, code in df[["road", "road_id"]]
        .drop_duplicates()
        .values
    }
}, "/content/traffic_model.joblib")

print("AI model trained and saved!")
print("Mean Absolute Error:", round(mae, 2), "km/h")
print("Model file exists:", os.path.exists("/content/traffic_model.joblib"))
```
